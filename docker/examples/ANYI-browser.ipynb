```python
import os
from pathlib import Path
from ppi_io import load_ppi

# In Docker, set PPI_DATA_DIR=/opt/ppi_app/data
DATA_DIR = Path(os.environ.get("PPI_DATA_DIR", "/opt/ppi_app/data"))

nodes_path = DATA_DIR / "nodes.pkl"
edges_path = DATA_DIR / "edges.csv"

result = load_ppi(
    nodes_path=nodes_path,
    edges_path=edges_path,
    node_col="node",
    u_col="source",
    v_col="target",
    graph_type="undirected",
    restrict_to_annotated_nodes=False,
)

df = result.nodes_df
edges = result.edges_df
G = result.graph

print("Annotation DataFrame shape:", df.shape)
print("Example columns:", df.columns.tolist()[:10])
print("Edge file columns:", edges.columns.tolist())
print(f"Graph has {G.number_of_nodes()} nodes and {G.number_of_edges()} edges")
print(f"Nodes with annotations & in graph: {result.n_overlap_annot_graph}")
```
