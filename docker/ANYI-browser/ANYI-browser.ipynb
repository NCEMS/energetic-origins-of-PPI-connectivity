# ANotated Yeast Interactome Browser

* Execute the code cell below to create a "Launch Browser" button.
* Click "Launch Browser" to initialize the interactive viewer

```python
import ipywidgets as W
from IPython.display import display, clear_output
from python_scripts.ppi_app import launch_app

out = W.Output()
btn = W.Button(description="Launch Browser", button_style="primary")

def on_click(_):
    with out:
        clear_output(wait=True)
        ui = launch_app(include_cytoscape=True)
        display(ui.root)

btn.on_click(on_click)
display(W.VBox([btn, out]))
```
