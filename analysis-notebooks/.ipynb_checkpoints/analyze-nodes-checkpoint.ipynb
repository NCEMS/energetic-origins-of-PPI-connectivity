### Annotated Yeast Interactome - introductory analysis notebook

* The annotated node file is available in the form of a pickled pd.DataFrame. 
* Storing the data in this format allows for the inclusion of data structures like lists and dictionaries directly in the file without fear of unintended changes to data
* [This document](https://docs.google.com/spreadsheets/d/145NyQs-qO5PaMc1dF3hRNpTDUANW4LKvrsjd3PAwl4Q/edit?usp=sharing) contains explanations of the meaning of the 133 column names in the pd.DataFrame
* **Notable missing pieces of information: translation speed / translation efficiency, LiP-MS data**

```python
import pandas as pd
import matplotlib.pyplot as plt

# this path is to the CyVerse data store - interacting with large files on the data store without transferring to the local file system
# will lead to problems, but we can read the annotated nodes file into memory without issue due to its small size (31 MB)

# path on CyVerse for most recent dataset
data_path = "/Users/dannissley/working-group-projects/energetic-origins/energetic-origins-of-PPI-connectivity/flatten/processed-data/3-s288c-nodes-final-per-node.pkl"

# path in ocal
#data_path = "Users/username/directory/path"
```

```python
nodes = pd.read_pickle(data_path)
print (f"Number of nodes in the full data file: {nodes.shape[0]}")
```

* This is the expected total number of nodes - the abstract for The Yeast Interactome (see the paper [here](https://www.nature.com/articles/s41586-023-06739-5)) mentions that 3,927 nodes are present

* Some nodes have strange names of the form `YER190W;YGR296W;YNL339C;YPL283C;YEL077C;YLL066C;YLL067C;YML133C;YPR204W`, making it difficult to unambiguously determine what sequence/structure should be used for analyses. These are filtered out below.

* Some nodes did not have a matching structure from AlphaFold2, so we will filter those out as well

```python
# APPLY FILTERS

# remove nodes with no sequence information from SGD
filtered_nodes = nodes[~nodes["sequence"].isna()]
print (f"Number of nodes with sequence information: {filtered_nodes.shape[0]}")

# record those nodes that do not have sequence information
no_seq_nodes = nodes[nodes["sequence"].isna()]
print (f"Number of nodes without sequence information: {no_seq_nodes.shape[0]}")

# remove nodes with no structure from AF2
filtered_nodes = filtered_nodes[filtered_nodes["structure_exists"] == 1]
print (f"Number of nodes with sequence & structure information: {filtered_nodes.shape[0]}")

# a small number of nodes have an AF2 structure, but its sequence does not match the sequence information from SGD - we can filter those out as well
filtered_nodes = filtered_nodes[filtered_nodes["sequence_matches_structure"] == True]
print (f"Number of nodes with sequence, structure, and matching information between the two: {filtered_nodes.shape[0]}")
```

### Network centrality information
* Information on network centralities is stored in the columns {`_wkshell`, `degree_centrality`, `betweenness_centrality`, `eigenvector_centrality`, `closeness_centrality`, `load_centrality`, `pagerank`, `information_centrality`, `CentralityCosDist_rank`, and `CentralityCosDist_similarity_score`}
* We can, for example filter to nodes with a `degree_centrality` in the top 5% with the below command
* Note that in this dataset, all centrality calculations were carried out on the full network with the exception of information centrality, for which I used the largest connected subgraph. 

```python
# compute the 95th percentile
threshold = filtered_nodes["degree_centrality"].quantile(0.95)

# filter by this value
top_5_percent = filtered_nodes[filtered_nodes["degree_centrality"] >= threshold]

# sort the result and print the first five rows
top_5_percent.sort_values(by="degree_centrality", ascending=False).head(5)
```

### Information on protein classification
* DeepTMHMM was used to classify proteins as transmembrane, secreted, and globular. We can see the class counts with the below command on the filtered dataset

```python
filtered_nodes["DeepTMHMM_class"].value_counts()
```

* Proteins classified as globular (`GLOB`) or signal peptide only (`SP`) were subjected to free energy calculations regardless of whether or not they contain predicted IDRs (see below)
* `BETA` = beta-barrel transmembrane protein
* `TM` = alpha-helical transmembrane protein
* `SP` = protein has a signal peptide but no transmembrane domain
* `SP+TM` = protein has a signal peptide and is membrane bound
* `GLOB` = protein has no signal peptide and is not membrane bound (presumably cytosolic)

### Intrinsically disordered regions
* metapredict v3.0 was used to predict IDRs within each protein
* The column `IDR_count` contains the integer number of IDRs containing at least 30 consecutive disordered residues for each protein

```python
filtered_nodes["IDR_count"].value_counts()
```

* The column `disorder_fraction` contains the proportion of residues within each protein predicted to be intrinsically disordered

```python
plt.clf()
plt.hist(filtered_nodes["disorder_fraction"], density=True, histtype="step")
plt.ylabel("Probability density")
plt.xlabel("Fraction disordered")
```

* The DisProt database entries for S288C yeast were used to classify proteins as overall disordered
* The columns {`is_disordered0.05`, `is_disordered0.10`, ..., `is_disordered0.95` `isdisordered1.00`} contain `1` if a given protein had a fraction of disordered residues greater than or equal to the {5th, 10th, ..., 95th, and 100th} percentile values of S288C proteins in DisProt and `0` if that protein has a smaller fraction of disordered residues

### IDR properties
* The properties of IDRs of at least 30 amino acids in length were computed using Cider (sequence properties) and Albatross (conformational ensemble properties)
* The columns `cider` and `albatross` contain nested dictionaries of these properties
* For example, below I extract one of these nested dictionaries to dispaly its structure and then plot a histogram of the predicted end to end distance across all predicted IDRs

```python
# extract the first node as an example
example = filtered_nodes.iloc[0]

# print the IDR sequences for this node
print ("IDR sequences:\n", example["IDR_sequences"], "\n")

# print the albatross dictionary for this node
print ("Albatross properties for these IDR sequences:\n", example["albatross"])
```

* The top level dictionary keys are IDR indices; these are matched to the sequences in `IDR_sequences`
* We can histogram all `idr_end_to_end` distance values with code like the following

```python
# extract the values
end_to_end_dist = []
for nested_dict in filtered_nodes["albatross"]:
    if isinstance(nested_dict, dict):
        for subdict in nested_dict.values():
            if isinstance(subdict, dict) and "idr_end_to_end" in subdict:
                end_to_end_dist.append(float(subdict["idr_end_to_end"]))

# plot the histogram
plt.clf()
plt.hist(end_to_end_dist, density=True, histtype="step", bins=20)
plt.ylabel("Probability density")
plt.xlabel("IDR end to end distance")
plt.xticks([0, 50, 100, 150, 200, 250, 300])
plt.xlim(-5, 305)
plt.ylim(-0.001, 0.025)
plt.show()
```

### UniProt and associated annotations
* Function, subcellular location, post-translational modification, protein name, and GO term data were extracted from UniProt
* In addition, the PTMeXchange phosphorylation dataset for yeast was integrated
* Below, I demonstrate how we can select all nodes with (1) a particular GO term or (2) containing a phosphotyrosine
* **Note about PTMeXchange data**: some yeast proteins are reported to have phosphorylated alanine residues; this appears to be a mass spec artifact. These records are included in the `.pkl` file. You may want to filter them out, but I have included them for completeness. 

```python
# find nodes with GO term "0005794", which indicates proteins associated with the golgi apparatus
GO_term = "0005794"
golgi_apparatus_nodes = filtered_nodes[filtered_nodes["GO_terms"].str.contains(GO_term, na=False)]
print (f"A total of {golgi_apparatus_nodes.shape[0]} nodes are associated with the golgi apparatus")
display(golgi_apparatus_nodes.head())

# find nodes that contain a phosphotyrosine; this may appear in either the parsed_PTMs column, which contains confirmed annotations within UniProt,
# or the ptm_gold, ptm_silver, and ptm_bronze columns, which contain information on phosphorylated residues from PTMeXchange mass spec data.
# Here, I limit the search over the PTMeXchange data to only "gold" results (highest confidence)
phosphoY_nodes = filtered_nodes[(filtered_nodes["parsed_PTMs"].str.contains("Phosphotyrosine", na = False)) |
                                (filtered_nodes["ptm_gold"].str.contains("Y", na = False))]

print(f"A total of {phosphoY_nodes.shape[0]} nodes contain a phosphotyrosine with high confidence")
display(phosphoY_nodes.head())
```

### Predicted post-translational modifications from PTMGPT2
* The presence and location of 19 classes of post-translational modifications were predicted with the model PTMGPT2
* You can locate nodes with, for example, *predicted* phosphotyrosines with code like the following

```python
pred_phosphoY_nodes = filtered_nodes[
    filtered_nodes["Phosphorylation (Y)"].notna() & 
    (filtered_nodes["Phosphorylation (Y)"].str.strip() != "")
]

print (f"A total of {pred_phosphoY_nodes.shape[0]} nodes are predicted to contain a phosphotyrosine by PTMGPT2")
display(pred_phosphoY_nodes.head())
```

### Free energy and structure-based scoring calculations
* Two free energey prediction methods (Ghosh & Dill 2010 Eq. 1 and Cagiada et al. 2024) were applied to non-transmembrane proteins
* The Cagiada model was trained with AF2 structures without relaxation beyond the 100 steps of steepest descent minimization applied in the standard AF2 relaxation protocol, and so I have run it on the AF2 structures as downloaded from EBI
* In addition to these methods, the Rosetta and FoldX scoring functions were used to compute various energy terms; Rosetta FastRelax was applied to structures before scoring with Rosetta ref2015 and FoldX
* These methods are highly sensitive to structure quality, and FastRelax in Rosetta will not globally refine the protein structure from a poor initial configuration
* Below, I apply a pLDDT threshold before plotting histograms of the results

*Note: signal sequence cleavage locations were predicted with SignalP6.0; all free energy calculations were carried out with the truncated sequence/structure. Later versions of this database will include protein structures predicted using the truncated sequences* 

```python
# filter to select globular proteins with high-quality AF2 structures
high_quality_struc = filtered_nodes[(filtered_nodes["mean_plddt"] > 0.9) & (filtered_nodes["DeepTMHMM_class"] == "GLOB")]

plt.figure(figsize=(10, 4))

plt.clf()
plt.subplot(1, 4, 1)
plt.hist(high_quality_struc["Ghosh-Dill-dG"], color="orange", density=True, histtype="step")
plt.title("Ghosh Dill dG")

plt.subplot(1, 4, 2)
plt.hist(-high_quality_struc["cagiada-dG"], color="dodgerblue", density=True, histtype="step")
plt.title("Cagiada dG")

plt.subplot(1, 4, 3)
plt.hist(high_quality_struc["Rosetta_total_score"], color="red", density=True, histtype="step")
plt.title("Rosetta total score")

plt.subplot(1, 4, 4)
plt.hist(high_quality_struc["FoldX_total_energy"], color="magenta", density=True, histtype="step")
plt.title("FoldX total energy")

plt.tight_layout()
```

### Some preliminary analyses
* The below is a preliminary analysis in which I compare, for the degree centrality metric, the top and bottom 5% of nodes on the basis of various properties
* *Note well*: this analysis is only intended to serve as an example of interacting with the pd.DataFrame object containing the annotated network of nodes

```python
from scipy.stats import mannwhitneyu
import numpy as np

# function to get the top5 and bottom5 % of rows based on a column name
def get_top_bottom_5_percent(df, column):
    top_5 = df.nlargest(int(0.05 * len(df)), column)
    bottom_5 = df.nsmallest(int(0.05 * len(df)), column)
    return top_5, bottom_5

# function to extract values for an IDR property
def extract_idr_ensemble_prop(df, prop_type, column):
    values = []
    for nested_dict in df[prop_type]:
        if isinstance(nested_dict, dict):
            for subdict in nested_dict.values():
                if isinstance(subdict, dict) and column in subdict:
                    values.append(float(subdict[column]))
    return values
```

```python
import matplotlib.pyplot as plt
from scipy.stats import mannwhitneyu
import numpy as np

# ensure t1/2 is numeric before slicing
filtered_nodes["t1/2 (min)"] = pd.to_numeric(filtered_nodes["t1/2 (min)"], errors="coerce")

# setup
centrality = "degree_centrality"
top5, bot5 = get_top_bottom_5_percent(filtered_nodes, centrality)
pvals = {}

metrics = {
    "cagiada-dG": "Cagiada ΔG",
    "Ghosh-Dill-dG": "Ghosh-Dill ΔG",
    "disorder_fraction": "Disorder Fraction",
    "t1/2 (min)": "Protein Half-life (min)",
    "Median molecules per cell": "Protein Abundance",
    "idr_rg_scaled": "Scaled IDR Radius of Gyration"
}

# create a 3x3 grid; leave middle top cell for bar plot
plt.clf()
fig, axes = plt.subplots(3, 3, figsize=(18, 12))
axes = axes.flatten()

bar_ax = axes[1]  # center-top of 3x3 grid

# compute statistics
means = [bot5[centrality].mean(), top5[centrality].mean()]
stds = [bot5[centrality].std(), top5[centrality].std()]
medians = [bot5[centrality].median(), top5[centrality].median()]
labels = ["Bottom 5%", "Top 5%"]

# draw bars
bars = bar_ax.bar(labels, means, yerr=stds, capsize=8, color=["blue", "orange"])

# annotate median next to each bar
for bar, median in zip(bars, medians):
    height = bar.get_height()
    bar_ax.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.01,
        f"median = {median:.3f}",
        ha="center",
        va="bottom",
        fontsize=10
    )

bar_ax.set_title("Mean Degree Centrality (±SD) with Median Labels")
bar_ax.set_ylabel("Degree Centrality")

# remove unused plots in top row
axes[0].axis('off')
axes[2].axis('off')

metric_axes = axes[3:]  

for ax, (col, title) in zip(metric_axes, metrics.items()):

    if col == "idr_rg_scaled":
        top_vals = extract_idr_ensemble_prop(top5, "albatross", col)
        bot_vals = extract_idr_ensemble_prop(bot5, "albatross", col)
    else:
        top_vals = top5[col].dropna().astype(float)
        bot_vals = bot5[col].dropna().astype(float)

    try:
        stat, p = mannwhitneyu(top_vals, bot_vals, alternative='two-sided')
        pvals[col] = p
    except ValueError:
        pvals[col] = np.nan

    # Draw boxplot
    ax.boxplot([bot_vals, top_vals], tick_labels=["Bottom 5%", "Top 5%"])

    # Get medians
    median_bot = np.median(bot_vals)
    median_top = np.median(top_vals)

    # Annotate medians above boxes
    ax.text(0.9, median_bot, f"median = {median_bot:.3f}", ha="center", va="bottom", fontsize=9, color="blue")
    ax.text(2.1, median_top, f"median = {median_top:.3f}", ha="center", va="bottom", fontsize=9, color="orange")

    ax.set_title(f"{title}\n(p = {p:.3e})")
    ax.set_ylabel("Value")

# Final layout
fig.suptitle(f"Comparison of Top vs Bottom 5% by {centrality}", fontsize=18)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
