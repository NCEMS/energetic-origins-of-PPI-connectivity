```python
import pandas as pd
import numpy as np
```

```python
# load the datasets with node centrality information and get nodes in both
DATA_PATH2 = "../flatten/processed-data/"
DATA_PATH3 = "../18-Y2H-data/processed-data/"
ms_nodes = pd.read_pickle(f"{DATA_PATH2}20251023-s288c-annotated-PPI-net.pkl")
y2h_nodes = pd.read_csv(f"{DATA_PATH3}Y2H-s288c-step18.csv")
```

```python
# determine which nodes are hubs based on quantile cutoff for degree centrality
cut = 0.10
ms_threshold_q = ms_nodes["degree_centrality"].quantile(1 - cut)
ms_hubs_q = ms_nodes[ms_nodes["degree_centrality"] >= ms_threshold_q]
ms_nothub_q  = ms_nodes[ms_nodes["degree_centrality"] <  ms_threshold_q]

y2h_threshold_q = y2h_nodes["degree_centrality"].quantile(1 - cut)
y2h_hubs_q   = y2h_nodes[y2h_nodes["degree_centrality"] >= y2h_threshold_q]
y2h_nothub_q = y2h_nodes[y2h_nodes["degree_centrality"] <  y2h_threshold_q]

print (f"Threshold for defining a hub in MS data is {ms_threshold_q:.5f}; there are {len(ms_hubs_q)} hubs and {len(ms_nothub_q)} non hubs")
print (f"Threshold for defining a hub in Y2H datais {y2h_threshold_q:.5f}; there are {len(y2h_hubs_q)} hubs and {len(y2h_nothub_q)} non hubs")
```

```python
# determine which nodes are hubs based on percentile cutoff for degree centrality
cut = 10
ms_threshold_p = np.percentile(ms_nodes["degree_centrality"], 100 - cut)
ms_hubs_p = ms_nodes[ms_nodes["degree_centrality"] >= ms_threshold_p]
ms_nothub_p  = ms_nodes[ms_nodes["degree_centrality"] <  ms_threshold_p]

y2h_threshold_p = np.percentile(y2h_nodes["degree_centrality"], 100 - cut)
y2h_hubs_p   = y2h_nodes[y2h_nodes["degree_centrality"] >= y2h_threshold_p]
y2h_nothub_p = y2h_nodes[y2h_nodes["degree_centrality"] <  y2h_threshold_p]

print (f"Threshold for defining a hub in MS data is {ms_threshold_p:.5f}; there are {len(ms_hubs_p)} hubs and {len(ms_nothub_p)} non hubs")
print (f"Threshold for defining a hub in Y2H datais {y2h_threshold_p:.5f}; there are {len(y2h_hubs_p)} hubs and {len(y2h_nothub_p)} non hubs")
```

```python
print (f"Number of matching hubs in MS data: {len(set(ms_hubs_p['node']) & set(ms_hubs_q['node']))}")
print (f"Number of matching non-hubs in MS data: {len(set(ms_nothub_p['node']) & set(ms_nothub_q['node']))}")
print (f"Number of matching hubs in the Y2H data: {len(set(y2h_hubs_p['node']) & set(y2h_hubs_q['node']))}")
print (f"Number of matching non-hubs in the Y2H data: {len(set(y2h_nothub_p['node']) & set(y2h_nothub_q['node']))}")
```
